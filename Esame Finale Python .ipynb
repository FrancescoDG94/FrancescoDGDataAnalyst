```python
# Aggregazione per continente
casi_per_continente = COVID_finale.groupby('continent').agg({
    'total_cases': 'sum',
    'location': 'count',
    'population': 'sum'
}).round(0)

casi_per_continente.columns = ['Casi_Totali', 'Num_Paesi', 'Popolazione_Totale']
casi_per_continente = casi_per_continente.astype({'Casi_Totali': 'int64', 'Num_Paesi': 'int64', 'Popolazione_Totale': 'int64'})
# Calcolo percentuali rispetto al mondo
totale_mondiale = casi_per_continente['Casi_Totali'].sum()
casi_per_continente['Percentuale_Mondiale'] = (casi_per_continente['Casi_Totali'] / totale_mondiale * 100).round(2)
# Calcolo casi per milione di abitanti
casi_per_continente['Casi_per_Milione'] = (casi_per_continente['Casi_Totali'] / casi_per_continente['Popolazione_Totale'] * 1000000).round(0).astype('int64')
# Ordinamento per numero di casi (decrescente)
casi_per_continente = casi_per_continente.sort_values('Casi_Totali', ascending=False)
# PRESENTAZIONE RISULTATI
print(f"TOTALE MONDIALE: {totale_mondiale:,} casi")
print("\n" + "=" * 80)
print(f"{'CONTINENTE':<15} {'CASI TOTALI':<15} {'% MONDO':<10} {'PAESI':<8} {'CASI/1M':<12}")
print("=" * 80)

for continente, row in casi_per_continente.iterrows():
    casi_formatted = f"{row['Casi_Totali']:,}"
    print(f"{continente:<15} {casi_formatted:<15} {row['Percentuale_Mondiale']:>6.2f}% {row['Num_Paesi']:>6} {row['Casi_per_Milione']:>10,}")

print("=" * 80)
```
